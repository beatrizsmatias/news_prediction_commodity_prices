```python
import os 
import requests 
import json
from urllib.parse import quote_plus
from newsdataapi import NewsDataApiClient
from bs4 import BeautifulSoup
```

```python
HEADERS = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64)"
}

def extract_full_article(url):
    try:
        res = requests.get(url, headers=HEADERS, timeout=10)
        if res.status_code == 200:
            soup = BeautifulSoup(res.text, "html.parser")

            # Tentativas comuns para pegar texto de artigos
            paragraphs = soup.find_all("p")
            text = " ".join([p.get_text() for p in paragraphs])

            # Limitar pra evitar texto muito grande (opcional)
            return text[:10000]  # Pega só os 3000 primeiros caracteres
        else:
            return "ERROR: Failed to fetch page."
    except Exception as e:
        return f"ERROR: {str(e)}"
```

# **NEWSDATA.IO**

```python
API_KEY = 'pub_ea04f645076a4547ad76594df2fb1c21'

client = NewsDataApiClient(apikey=API_KEY)

def split_keywords(keywords, max_len=100):
    batches = []
    current_batch = []
    current_len = 0
    for kw in keywords:
        # conta o tamanho com ' OR ' extra se não for o primeiro
        addition = len(kw) + (4 if current_batch else 0)
        if current_len + addition > max_len:
            batches.append(current_batch)
            current_batch = [kw]
            current_len = len(kw)
        else:
            current_batch.append(kw)
            current_len += addition
    if current_batch:
        batches.append(current_batch)
    return batches

all_news = []

batches = split_keywords(keywords)

for batch in batches:
    query = " OR ".join(batch)
    print(f"Buscando por: {query}")
    try:
        response = client.latest_api(q=query, language='en')
        results = response.get('results', [])
        for article in results:
            link = article.get("link", "")
            noticia = {
                "title": article.get('title'),
                "category": article.get('category', ['General'])[0] if isinstance(article.get('category'), list) else article.get('category', 'General'),
                "summary": article.get('description'),
                "link": article.get('link'),
                "publication_time": article.get('pubDate'),
                "full_article": extract_full_article(link)

            }
            all_news.append(noticia)
    except Exception as e:
        print(f"Erro ao buscar batch: {e}")

print(json.dumps(all_news, indent=4, ensure_ascii=False))
```

# **LIMPA NOTICIAS**

```python
import json
import re

# Configuração
input_file = "news.json"
output_file = "news_2.json" MUDAR AQUI O NUMERO DA NOTICIA 

# Palavras-chave e padrões que indicam NÃO NOTÍCIA
LIXO_PADROES = [
    r"(?i)please register",
    r"(?i)sign in",
    r"(?i)sign up",
    r"(?i)subscribe",
    r"(?i)agree & join",
    r"(?i)create (your )?account",
    r"(?i)this material may not be published",
    r"(?i)terms of use",
    r"(?i)privacy policy",
    r"(?i)complete digital access",
    r"(?i)legal statement",
    r"(?i)thank you for visiting",
    r"(?i)cookies? policy",
    r"(?i)advertisement",
    r"(?i)join now",
    r"(?i)to view this content",
    r"(?i)your email address"
]

# Regex para detectar conteúdo muito técnico SEM contexto de evento
# (exemplo: EOR puro, metodologia, sem "reports", "said", "announced"...)
TECHNICAL_ONLY = [
    r"(?i)enhanced oil recovery",
    r"(?i)surfactant",
    r"(?i)IFT",
    r"(?i)wettability",
    r"(?i)nanofluid",
    r"(?i)FTIR",
    r"(?i)x-ray fluorescence",
    r"(?i)thermogravimetric"
]

# Verbos que indicam narrativa factual — se NÃO tiver, pode ser técnico
VERBOS_FATUAIS = ["report", "said", "announced", "signed", "warned", "confirmed", "plans", "expected"]

# Abrir arquivo original
with open(input_file, "r", encoding="utf-8") as f:
    data = json.load(f)

limpas = []

for item in data:
    texto = item.get("full_article", "").strip().lower()
    # 1️⃣ Se vazio: descarta
    if not texto:
        continue

    # 2️⃣ Se começar ou contiver padrões de lixo: descarta
    if any(re.search(padrao, texto) for padrao in LIXO_PADROES):
        continue

    # 3️⃣ Se contiver termos técnicos mas não tem verbo factual => descarta
    if any(re.search(padrao, texto) for padrao in TECHNICAL_ONLY):
        if not any(v in texto for v in VERBOS_FATUAIS):
            continue

    # 4️⃣ Se muito curto e sem verbo factual => descarta
    if len(texto) < 200:
        if not any(v in texto for v in VERBOS_FATUAIS):
            continue

    # Passou no filtro? Mantém!
    limpas.append(item)

# Salvar resultado final
with open(output_file, "w", encoding="utf-8") as f:
    json.dump(limpas, f, ensure_ascii=False, indent=4)

print(f"✅ JSON limpo salvo em: {output_file} ({len(limpas)} notícias finais)")
```
